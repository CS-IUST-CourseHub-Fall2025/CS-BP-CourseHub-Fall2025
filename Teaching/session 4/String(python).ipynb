# Introduction to Strings in Python

A string in Python is a sequence of characters used to store text data, such as words or sentences.
Strings are written inside single or double quotes, and each character in a string has an index starting from 0.
They are one of the most common data types in Python and are widely used for handling and displaying text.

## 1. Single and Double Quotation
In Python, strings can be created using either single quotes (' ') or double quotes (" "). Both work the same way.
This flexibility helps when your string contains one type of quote — for example, you can use double quotes to wrap a string that contains single quotes.

Example:

```python
a = 'Hello World'
b = "Hello World"
c = "It's a beautiful day"
print(a)
print(b)
print(c)
```

## 2. Multiline Strings

If you want to store a string that spans multiple lines, you can use triple quotes (''' or """).
This is useful for paragraphs, documentation, or formatted text.

Example:

```python
text = '''This is a
multiline
string.'''
print(text)
```

## 3. Strings are Arrays
In Python, strings are arrays of characters. You can access specific characters using their index number.
Remember that indexing starts from 0.

Example:

```python
a = "Hello"
print(a[2])  # Output: 'l'
print(a[-1])  # Output: 'o'
```

## 4. String Length

To find the length of a string, use the built-in len() function.

Example:

```python
a = "Hello World"
print(len(a))  # Output: 11
```

## 5. Checking if a String Contains a Letter

You can use the in keyword to check if a substring exists within a string. It returns True or False.

Example:

```python
a = "Hello World"
print("H" in a)  # True
print("z" in a)  # False
```

## 6. Slicing Strings
Slicing allows you to extract a portion of a string using the syntax string[start:end].
Note that the end index is not included in the result.

Example:

```python
a = "Hello World"
print(a[0:5])   # 'Hello'  → characters from index 0 to 4
print(a[6:11])  # 'World'  → characters from index 6 to 10
print(a[:5])    # 'Hello'   → from start to index 4
print(a[6:])    # 'World'   → from index 6 to the end
print(a[1:-2])   # 'ello Wor'  → starts from index 1 ('e') and goes up to two characters before the end ('r')
```

## 7. Modify Strings

Python provides many built-in methods to modify strings:

upper(): Converts all characters to uppercase.

lower(): Converts all characters to lowercase.

strip(): Removes whitespace from the beginning and end.

replace(): Replaces parts of a string with another.

split(): Splits a string into a list.

Example:

```python
a = " Hello World  "
print(a.upper())           # HELLO WORLD
print(a.lower())           # hello world
print(a.strip())           # Hello World
print(a.replace("World", "Python"))  # Hello Python
print(a.split())           # ['Hello', 'World']
```

## 8. String Concatenation
You can combine strings using the + operator. Remember to include spaces if needed.

Example:

```python
first_name = "John"
last_name = "Doe"
full_name = first_name + " " + last_name
print(full_name)  # Output: John Doe

announcement = "Welcome"
announcement += " " + full_name + "!"
print(announcement)  # Output: Welcome John Doe!

age = 30
# print("I am " + age)  # TypeError
print("I am " + str(age))  # Output: I am 30
```

## 9. f-Strings and format()
To insert variables inside strings, you can use f-strings (formatted string literals) or the format() method.
f-strings are available in Python 3.6 and above.

Example:

```python
name = "Amir"
age = 20
print(f"My name is {name}, and I am {age} years old.")
print("My name is {}, and I am {} years old.".format(name, age))
```

## 10. Escape Characters

Escape characters allow you to include special characters in strings. Common examples include:

\n → New line

\t → Tab space

\\\ → Backslash

```python
text = "Hello\nWorld\tPython"
print(text)
```

## 11. String Methods

Python provides many useful string methods. Some common ones include:

count():    Counts occurrences of a substring.

find(): Returns the index of the first occurrence of a substring.

index():    Similar to find(), but raises an error if the substring is not found.

split():    Splits a string into a list.

is⚪(): isalpha(), isdigit(), islower(), isupper(), etc. check string characteristics.

Example:

```python
a = "Hello World"
print(a.count("l"))   # 3
print(a.find("o"))    # 4
print(a.index("W"))   # 6
print(a.split())      # ['Hello', 'World']
print(a.isalpha())    # False (because there's a space)
```
