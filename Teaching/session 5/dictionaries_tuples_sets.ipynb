## Dictionaries

### 1. Dictionary example

```python
# Multiple ways to create dictionaries
a = dict(one=1, two=2, three=3)
b = {'one': 1, 'two': 2, 'three': 3}
c = dict(zip(['one', 'two', 'three'], [1, 2, 3]))
d = dict([('two', 2), ('one', 1), ('three', 3)])
e = dict({'three': 3, 'one': 1, 'two': 2})
f = dict({'one': 1, 'three': 3}, two=2)
print(a)
```

### 2. Accessing Dictionary Items

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
print(thisdict["model"])  # Mustang
print(thisdict.get("year"))  # 1964
# Checking membership
print("brand" in thisdict)  # True
```

### 3. Changing Dictionary Items

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
# Change value of an existing key
thisdict["year"] = 2018
print(thisdict)  # {'brand': 'Ford', 'model': 'Mustang', 'year': 2018}

# Using update() to change an existing key
thisdict.update({"year": 2020})
print(thisdict)  # {'brand': 'Ford', 'model': 'Mustang', 'year': 2020}
```

### 4. Adding Dictionary Items

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
# Add a new key-value pair
thisdict["color"] = "red"
print(thisdict)  # {'brand': 'Ford', 'model': 'Mustang', 'year': 1964, 'color': 'red'}

# Using update() to add a new key-value pair
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
thisdict.update({"color": "white"})
print(thisdict)  # {'brand': 'Ford', 'model': 'Mustang', 'year': 1964, 'color': 'white'}
```

### 5. Removing Dictionary Items

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
# Remove item with pop()
thisdict.pop("model")
print(thisdict)  # {'brand': 'Ford', 'year': 1964}

thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
# Remove item with popitem() (removes last inserted item)
thisdict.popitem()
print(thisdict)  # {'brand': 'Ford', 'model': 'Mustang'}

thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
# Remove item using del
del thisdict["model"]
print(thisdict)  # {'brand': 'Ford', 'year': 1964}

thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
# Clear all items
thisdict.clear()
print(thisdict)  # {}

# Delete entire dictionary
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
del thisdict
# print(thisdict)  # NameError: name 'thisdict' is not defined
```

### 6. Loop and iterating through Dictionaries

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
# Loop through keys
for key in thisdict:
    print(key)  # brand, model, year

# Loop through values via keys
for key in thisdict:
    print(thisdict[key])  # Ford, Mustang, 1964

# Loop through values directly
for value in thisdict.values():
    print(value)  # Ford, Mustang, 1964

# Loop through keys explicitly
for key in thisdict.keys():
    print(key)  # brand, model, year

# Loop through key-value pairs
for key, value in thisdict.items():
    print(key, value)  # brand Ford, model Mustang, year 1964
```

### 7. Copy Dictionaries

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
# Using copy()
mydict = thisdict.copy()
print(mydict)  # {'brand': 'Ford', 'model': 'Mustang', 'year': 1964}

# Using dict() constructor
mydict2 = dict(thisdict)
print(mydict2)  # {'brand': 'Ford', 'model': 'Mustang', 'year': 1964}
```

### 8. Nested Dictionaries

```python
# Nested dictionary example
myfamily = {
    "child1": {"name": "Emil", "year": 2004},
    "child2": {"name": "Tobias", "year": 2007},
    "child3": {"name": "Linus", "year": 2011}
}
print(myfamily)

# Another way to create nested dictionary
child1 = {"name": "Emil", "year": 2004}
child2 = {"name": "Tobias", "year": 2007}
child3 = {"name": "Linus", "year": 2011}
myfamily = {"child1": child1, "child2": child2, "child3": child3}
print(myfamily)

# Access nested value
print(myfamily["child2"]["name"])  # Tobias

# Loop through nested dictionary
for child, info in myfamily.items():
    print(child, info)
    for key in info:
        print(key + ":", info[key])
```

### 9. All Dictionary methods

```python
d = {"a": 1, "b": 2, "c": 3}
# clear()
d2 = d.copy()
d2.clear()
print(d2)  # {}

# copy()
print(d.copy())  # {'a': 1, 'b': 2, 'c': 3}

# fromkeys()
keys = ('x', 'y', 'z')
print(dict.fromkeys(keys, 0))  # {'x': 0, 'y': 0, 'z': 0}
print(dict.fromkeys(keys))    # {'x': None, 'y': None, 'z': None}

# get()
print(d.get('a'), d.get('z', 0))  # 1 0

# items()
print(d.items())  # dict_items([('a', 1), ('b', 2), ('c', 3)])

# keys()
print(d.keys())   # dict_keys(['a', 'b', 'c'])

# pop()
d2 = {"a":1, "b":2}
print(d2.pop('a'), d2)  # 1 {'b': 2}

# popitem()
d3 = {"a":1, "b":2}
item = d3.popitem()
print(item, d3)  # ('b', 2) {'a': 1}

# setdefault()
d4 = {"a":1}
print(d4.setdefault('b', 3), d4)  # 3 {'a': 1, 'b': 3}
print(d4.setdefault('a', 9))      # 1 (existing value)

# update()
d5 = {"a":1}
d5.update({"b":2})
print(d5)  # {'a': 1, 'b': 2}

# values()
print(d.values())  # dict_values([1, 2, 3])
```

### 10. Dictionary Comprehension

```python
# Simple dict comprehension
nums = [1, 2, 3, 4, 5]
squares = {x: x**2 for x in nums}
print(squares)  # {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}

# Dictionary comprehension with if condition
even_squares = {x: x**2 for x in range(10) if x % 2 == 0}
print(even_squares)  # {0: 0, 2: 4, 4: 16, 6: 36, 8: 64}

# Comprehension from tuple and list
keys = ['a','b','c']
values = [1,2,3]
mapping = {k: v for (k, v) in zip(keys, values)}
print(mapping)  # {'a': 1, 'b': 2, 'c': 3}

# Nested comprehension example
letters = ['x','y']
numbers = [1,2]
nested = {letter: {num: letter+str(num) for num in numbers} for letter in letters}
print(nested)  # {'x': {1: 'x1', 2: 'x2'}, 'y': {1: 'y1', 2: 'y2'}}
```

## Tuples

### 1. Tuple example

```python
# Basic tuple creation
thistuple = ("apple", "banana", "cherry")
print(thistuple)  # ('apple', 'banana', 'cherry')

# Tuples can have duplicates and different data types
thistuple = ("apple", "banana", "cherry", "apple", "cherry")
print(thistuple)  # duplicates allowed
print(len(thistuple))  # 5

# One-item tuple (note the comma)
single = ("apple",)
print(type(single))  # <class 'tuple'>

# Using tuple() constructor
thistuple = tuple(("apple", "banana", "cherry"))
print(thistuple)  # ('apple', 'banana', 'cherry')
```

### 2. Accessing Tuples and Slicing of Tuple  

```python
thistuple = ("apple", "banana", "cherry", "orange", "kiwi", "melon")
print(thistuple[1])    # banana
print(thistuple[-1])   # melon

# Slicing
print(thistuple[2:5])  # ('cherry', 'orange', 'kiwi')
print(thistuple[:4])   # ('apple', 'banana', 'cherry', 'orange')
print(thistuple[2:])   # ('cherry', 'orange', 'kiwi', 'melon')
print(thistuple[-4:-1])# ('orange', 'kiwi', 'melon')

# Check if item exists
if "apple" in thistuple:
    print("Yes, 'apple' is in the tuple")
```

### 3. Updating Tuples and Deleting a Tuple

```python
thistuple = ("apple", "banana", "cherry")
# Convert to list to modify
y = list(thistuple)
y[1] = "kiwi"
thistuple = tuple(y)
print(thistuple)  # ('apple', 'kiwi', 'cherry')

# Adding items by converting to list or concatenation
thistuple = ("apple", "banana", "cherry")
y = list(thistuple)
y.append("orange")
thistuple = tuple(y)
print(thistuple)  # ('apple', 'banana', 'cherry', 'orange')

# Or using concatenation (must add comma)
thistuple = ("apple", "banana", "cherry")
thistuple += ("orange",)
print(thistuple)  # ('apple', 'banana', 'cherry', 'orange')

# Removing items
thistuple = ("apple", "banana", "cherry")
y = list(thistuple)
y.remove("apple")
thistuple = tuple(y)
print(thistuple)  # ('banana', 'cherry')

# Deleting tuple
thistuple = ("apple", "banana", "cherry")
del thistuple
# print(thistuple)  # NameError: tuple is deleted
```

### 4. Unpack Tuples

```python
fruits = ("apple", "banana", "cherry")
# Packing a tuple:
packed = fruits
print(packed)  # ('apple', 'banana', 'cherry')

# Unpacking a tuple:
green, yellow, red = fruits
print(green)   # apple
print(yellow)  # banana
print(red)     # cherry

# Using * to collect extra values
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")
green, yellow, *red = fruits
print(green)  # apple
print(yellow) # banana
print(red)    # ['cherry', 'strawberry', 'raspberry']

# Asterisk in middle
fruits = ("apple", "mango", "papaya", "pineapple", "cherry")
green, *tropic, red = fruits
print(green)  # apple
print(tropic) # ['mango', 'papaya', 'pineapple']
print(red)    # cherry
```

### 5. Loop and iterating through Tuples

```python
thistuple = ("apple", "banana", "cherry")
# Using for loop
for x in thistuple:
    print(x)

# Using index with range()
for i in range(len(thistuple)):
    print(thistuple[i])

# Using while loop
i = 0
while i < len(thistuple):
    print(thistuple[i])
    i += 1
```

### 6. Join Tuples

```python
tuple1 = ("a", "b", "c")
tuple2 = (1, 2, 3)
# Join with +
tuple3 = tuple1 + tuple2
print(tuple3)  # ('a', 'b', 'c', 1, 2, 3)

# Multiply tuple
fruits = ("apple", "banana", "cherry")
mytuple = fruits * 2
print(mytuple)  # ('apple','banana','cherry','apple','banana','cherry')
```

### 7. All Tuple methods

```python
thistuple = (1, 2, 3, 2, 2)
print(thistuple.count(2))  # 3
print(thistuple.index(3))  # 2
```

## Sets

### 1. Set example

```python
# Create a set
thisset = {"apple", "banana", "cherry"}
print(thisset)  # {'banana', 'apple', 'cherry'} (unordered)

# Duplicates are ignored
thisset = {"apple", "banana", "cherry", "apple"}
print(thisset)  # {'banana', 'apple', 'cherry'}

# Mixed types
set1 = {"abc", 34, True, 40, "male"}
print(set1)

# Using set() constructor
thisset = set(("apple", "banana", "cherry"))
print(thisset)
```

### 2. Accessing Set Items

```python
thisset = {"apple", "banana", "cherry"}
# Looping through set
for x in thisset:
    print(x)

# Check for membership
print("banana" in thisset)  # True
```

### 3. Adding Set Items

```python
thisset = {"apple", "banana", "cherry"}
# Add an item with add()
thisset.add("orange")
print(thisset)  # {'banana', 'orange', 'apple', 'cherry'}

# Add multiple items with update()
thisset = {"apple", "banana", "cherry"}
tropical = {"pineapple", "mango", "papaya"}
thisset.update(tropical)
print(thisset)  # {'banana', 'papaya', 'apple', 'pineapple', 'mango', 'cherry'}

# Add elements of a list
thisset = {"apple", "banana", "cherry"}
mylist = ["kiwi", "orange"]
thisset.update(mylist)
print(thisset)  # {'banana', 'kiwi', 'apple', 'orange', 'cherry'}
```

### 4. Removing Set Items

```python
thisset = {"apple", "banana", "cherry"}
# Remove with remove()
thisset.remove("banana")
print(thisset)  # {'apple', 'cherry'}

thisset = {"apple", "banana", "cherry"}
# Remove with discard()
thisset.discard("banana")
print(thisset)  # {'apple', 'cherry'}

# Remove with pop() (random)
thisset = {"apple", "banana", "cherry"}
x = thisset.pop()
print(x)       # random item (e.g. 'apple')
print(thisset) # remaining set

# Clear set
thisset = {"apple", "banana", "cherry"}
thisset.clear()
print(thisset)  # set()

# Delete set
thisset = {"apple", "banana", "cherry"}
del thisset
# print(thisset)  # NameError: thisset no longer exists
```

### 6. Join Sets

```python
# Union of sets (new set)
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = set1.union(set2)
print(set3)  # {'a', 1, 2, 3, 'b', 'c'}

# Using | operator
set3 = set1 | set2
print(set3)  # {'a', 1, 2, 3, 'b', 'c'}

# Update (adds set2 into set1)
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set1.update(set2)
print(set1)  # {'a', 1, 2, 3, 'b', 'c'}

# Intersection
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set3 = set1.intersection(set2)
print(set3)  # {'apple'}

# Using & operator
set1 = {"apple", "banana", "cherry"}
set3 = set1 & set2
print(set3)  # {'apple'}

# Intersection_update (modify set1)
set1 = {"apple", "banana", "cherry"}
set1.intersection_update(set2)
print(set1)  # {'apple'}
```

### 7. Frozenset

```python
# Frozenset is immutable
x = frozenset({"apple", "banana", "cherry"})
print(x)            # frozenset({'banana', 'apple', 'cherry'})
print(type(x))      # <class 'frozenset'>
```

### 8. All Set methods

```python
set1 = {1, 2, 3}
set2 = {2, 3, 4}
# difference()
print(set1.difference(set2))    # {1}

# intersection()
print(set1.intersection(set2))  # {2, 3}

# isdisjoint()
print(set1.isdisjoint({5,6}))   # True

# issubset() / issuperset()
print({1,2}.issubset(set1))     # True
print(set2.issuperset({3}))     # True

# symmetric_difference()
print(set1.symmetric_difference(set2))  # {1, 4}

# union()
print(set1.union(set2))         # {1, 2, 3, 4}

# pop(), remove(), discard() covered above
# clear() covered above
```
