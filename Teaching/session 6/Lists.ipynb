## **1. Defining Lists**
- Lists are used to stora multiple variables in a single variable.

- Lists are created with square brackets in python:

```python
lst = ["apple", 2, "A"]
print(lst)
```

## **2. List features**
- List items are ordered, changeable, and allow duplicate values.

- List items are indexed, the first item has index [0], the second item has index [1] etc

- List items are changeable

- The list is changeable, meaning that we can change, add, and remove items in a list after it has been created.

- Since lists are indexed, they can have a duplicate value e.g

```python
lst = ["car", "hello", "2", "car"]
print(lst)
```

## **4. List access**

- List items are indexed and you can access them by referring to the index number:
- Negative index means counting from the end.
- You can also specify a range of indexes by defining where to start and where to end.

```python
lst = [1, 2, 3, 4, 5]
print(lst[1])
print(lst[-1])
print(lst[2:])
print(lst[1:4])
```

## **5. Change Items Value** 
- You can change the value of an item in the list by reffering its index number.
- To change the value of items within a specific range, define a list with the new values, and refer to the range of index numbers where you want to insert the new values.

## Adding, Removing, and Modifying Elements

- Add → `append()`, `insert()`, `extend()`
- Remove → `remove()`, `pop()`, `clear()`
- Modify → reassign by index

```python
fruits = ["apple", "banana", "cherry"]

# Add
fruits.append("orange")
fruits.insert(1, "kiwi")

# Remove
fruits.remove("banana")
last = fruits.pop()

# Modify
fruits[0] = "grape"

print(fruits)
```

## **6. Useful List Methods**

| Method | Description |
|:-------|:-------------|
| `append(x)` | Add element to end |
| `extend(iterable)` | Add multiple elements |
| `insert(i, x)` | Insert element at index |
| `remove(x)` | Remove first occurrence |
| `pop(i)` | Remove and return element |
| `clear()` | Remove all items |
| `index(x)` | Return first index of x |
| `count(x)` | Count occurrences of x |
| `sort()` | Sort list |
| `reverse()` | Reverse list |
| `copy()` | Shallow copy |

```python
nums = [4, 1, 3, 2]
nums.sort()
print(nums)
print(nums.pop(0))
print(nums.count(5))

additional = [5, 6, 7]
nums.extend(additional)
nums.reverse()
print(nums)
```

## **7. Copying Lists (Important!)**

⚠️ Copying lists incorrectly can cause unexpected behavior.

```python
a = [1, 2, 3]
b = a          # Not a real copy – both point to the same list
c = a.copy()   # Independent copy

a.append(4)
print("a =", a)
print("b =", b)
print("c =", c)
```

## **8. List Comprehension**

List comprehensions provide a concise way to create new lists from existing iterables .

The basic syntax is:

new_list = [expression for item in iterable if condition]

- For example, suppose you have a list of fruits and want a new list containing only those fruits with the letter
"a" :

```python

fruits = ["apple", "banana", "cherry", "kiwi", "mango"]
newlist = [x for x in fruits if "a" in x]
print(newlist) # Output: ['apple', 'banana', 'mango']
```

This is equivalent to writing a loop with an if statement, but more compact . Without the condition,

you create a copy of the list or apply a transformation to each element:


```python
# Copy all items
newlist = [x for x in fruits]
# Or square each number in a list
numbers = [1, 2, 3, 4, 5]
squares = [x**2 for x in numbers]
print(squares) # Output: [1, 4, 9, 16, 25]
```

You can also include an else expression or more complex conditions in a comprehension. But at its core,

list comprehension is about generating a new list in a single, readable line .

## **9. Loop Lists**


- *You can iterate over list elements using loops. The simplest is a for loop:*



```python
thislist = ["apple", "banana", "cherry"]
for fruit in thislist:
    print(fruit)

#This will print each item in the list one by one . You can also loop by index using range(len(list))

thislist = ["apple", "banana", "cherry"]
for i in range(len(thislist)):
    print(thislist[i])
```

Using range(len(list)) produces indices [0, 1, ..., len(list)-1] , so thislist[i]

accesses each element. A while loop can also be used to iterate by index if needed. The loop does not

modify the list; it simply reads each element in order.

### **10. Enumerate (Loop with Index)**

When looping through a list, sometimes you need both the item and its index number. `enumerate` provides this capability.

```python
project_team = ["Ehsan", "Amir", "Reza"]

for index, name in enumerate(project_team):
    print(f"Student #{index + 1} is {name}")
```

### **11. Join List to String**

The `.join()` method allows you to combine a list of strings into a single string, separated by a delimiter.

```python
message_words = ["Welcome", "to", "IUST", "University"]
sentence = " ".join(message_words)
print(sentence)
```

### **12. Check Existence (in)**

Use the `in` keyword to check if a value exists inside a list. It returns `True` or `False`.

```python
passed_students = ["Amin", "Sara", "Amir"]

if "Amin" in passed_students:
    print("Amin has passed.")

if "Reza" not in passed_students:
    print("Reza is not in the list.")
```
