## Try/Except Basics

`try` and `except` blocks let you handle runtime errors without crashing the program. The code inside a `try:` block is executed normally unless an exception occurs; if an exception is raised, control jumps to the matching `except` block. For example:

```python
# Example 1: basic try/except
try:
    result = 10 // 2
    print("Division result:", result) # Division result: 5
except:
    print("An error occurred")
```

If an error occurs in the `try` block, the `except` block runs. In this example, no error occurs, so the `except` block is skipped. If we divide by zero:

```python
# Example 2: catching ZeroDivisionError
try:
    result = 10 // 0
    print("Division result:", result)
except ZeroDivisionError:
    print("Cannot divide by zero") # Output: Cannot divide by zero
```

A bare `except:` (no exception type) will catch any exception, but it’s often better to catch specific exceptions.

## 1. Catching Specific Exceptions

You can specify exception types to catch using `except ErrorType:`. This lets you handle different errors separately. For example:

```python
# Example 3: catching different exception types
try:
    x = int("abc") # causes ValueError
except ValueError:
    print("ValueError: invalid literal for int()") # This will print
except TypeError:
    print("TypeError: incompatible types")
```

If the code in `try` raises a `ValueError`, the first `except` runs; a `TypeError` would trigger the second. You can also use multiple `except` blocks for different errors:

```python
try:
    print(unknown_variable) # NameError: name 'unknown_variable' is not defined
except NameError:
    print("NameError: variable is not defined") # This will print
except Exception as e:
    print("Caught an exception:", e)
```

Here the first `except` handles `NameError`, and the second is a general catch-all (`Exception`).

## 2. Catching Multiple Exceptions

You can catch several exception types in one `except` by using a tuple:

```python
# Example 4: catching multiple exceptions in one block
try:
    y = [1, 2, 3]
    value = y[5] # IndexError
except (IndexError, KeyError) as e:
    print("Index/Key error:", e) # Output: Index/Key error: list index out of range
```

This catches either an `IndexError` or `KeyError`. You can also catch all exceptions with `except Exception as e:` and inspect `e`.

## 3. The else Clause

An optional `else:` block runs only **if no exception was raised in the try block**. Use it for code that should run when everything succeeds. For example:

```python
# Example 5: using else clause
try:
    res = (4 + 3) // (7 - 5)
except ZeroDivisionError:
    print("Division by zero occurred")
else:
    print("Success:", res) # Output: Success: 3
```

If no exception occurs, the `else` block prints the result. If we change inputs to cause a zero division, the `except` runs instead:

```python
try:
    res = (4 + 3) // (7 - 7)
except ZeroDivisionError:
    print("Division by zero occurred") # This will print
else:
    print("Success:", res)
```

Using `else` avoids putting extra code inside `try` and ensures the `except` only catches errors from the intended code.

## 4. The finally Clause

An optional `finally:` block will execute **no matter what** – whether an exception occurred or not. It’s typically used for clean-up actions. For example:

```python
# Example 6: using finally
try:
    val = 5 // 0
    print(val)
except ZeroDivisionError:
    print("Caught division by zero") # Output: Caught division by zero
finally:
    print("Cleanup actions (always executes)")
# Output: Cleanup actions (always executes)
```

Even though an exception occurred, the `finally` block still executes. If no exception occurs, `finally` still runs:

```python
try:
    val = 5 // 1
    print("Result:", val) # Output: Result: 5
except ZeroDivisionError:
    print("Caught division by zero")
finally:
    print("Cleanup actions (always executes)")
# Output: Cleanup actions (always executes)
```

If there is a `return`, `break`, or `continue` in `try` or `except`, the `finally` block runs just before those actions complete.

## 5. Nested Try-Except

You can nest `try` / `except` blocks to handle exceptions at multiple levels. For example:

```python
# Example 7: nested try/except
def process(a, b):
    try:
        result = a // b
        print("Result:", result)
        try:
            data = [1, 2, 3]
            print("Accessing index 5:", data[5]) # raises IndexError
        except IndexError:
            print("Inner except: index out of range") # This will print
    except ZeroDivisionError:
        print("Outer except: cannot divide by zero")
    finally:
        print("Always executes at end")

print("--- Test 1 ---")
process(6, 2)
# Output:
# Result: 3
# Inner except: index out of range
# Always executes at end

print("\n--- Test 2 ---")
process(6, 0)
# Output:
# Outer except: cannot divide by zero
# Always executes at end
```

In this example, dividing by zero triggers the outer `except`. Accessing an invalid list index triggers the inner `except`. The `finally` block runs after either case.

## Raising Exceptions

In addition to catching exceptions, you can raise them manually using the `raise` statement. Use this to indicate error conditions:

```python
# Example 8: raising an exception
try:
    x = -1
    if x < 0:
        raise ValueError("Negative value not allowed")
except ValueError as e:
    print("Raised error:", e) 
    # Output: Raised error: Negative value not allowed
```

This `raise` causes a `ValueError` to be thrown. The program jumps to the `except` block where it is caught and handled.

You can also re-raise an exception to let it propagate after some handling. For example:

```python
# Example 9: catching and re-raising an exception
try:
    try:
        raise NameError("Inner NameError")
    except NameError:
        print("Caught NameError inside, re-raising")
        raise
except NameError as e:
    print("Outer catch:", e) 
    # Output: Outer catch: Inner NameError
```

Here, the inner `except` catches `NameError`, prints a message, and then uses `raise` with no arguments to re-raise the same exception. The outer block then catches it. The official documentation notes that using `raise` without arguments inside an `except` re-raises the current exception.

You can also use `raise ... from ...` to chain exceptions, preserving the original context:

```python
# Example 10: exception chaining
try:
    open("nonexistent_file.txt")
except FileNotFoundError as e:
    raise RuntimeError("File handling failed") from e
```

This produces a `RuntimeError` whose cause is the original `FileNotFoundError`, as shown in the traceback.

### Key Points:
* Every `try` can have multiple `except` blocks, one `else`, and one `finally`.
* The `else` block runs only **if no exceptions occur**.
* The `finally` block **always runs**.
* Use specific exceptions when possible, and use `raise` to signal error conditions explicitly.

These features allow writing robust Python code that handles errors gracefully.
