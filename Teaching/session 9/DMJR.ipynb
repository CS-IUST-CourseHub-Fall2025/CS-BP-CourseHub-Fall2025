# Dates in Python

Python has a built-in module called datetime to work with dates and times. You can perform operations such as creating dates, formatting them, and extracting specific parts.

```python
import datetime

# Get today's date
today = datetime.date.today()
print("Today's date:", today)
```

### 1. Date Output
You can access year, month, and day individually:

```python
print("Year:", today.year)
print("Month:", today.month)
print("Day:", today.day)
```

### 2. Creating Date Objects
You can create a custom date using datetime.date(year, month, day):

```python
birthday = datetime.date(2006, 8, 28)
print("Birthday:", birthday)
```

### 3. The strftime() Method
The strftime() method formats dates as strings. Common format codes:

%Y → Year (4 digits)

%m → Month (01-12)

%d → Day (01-31)

%A → Weekday full name

%B → Month full name

```python
today = datetime.date(2025, 12, 1)

# Individual demonstrations
print("Weekday (%A):", today.strftime("%A"))       # Full weekday name
print("Month (%B):", today.strftime("%B"))         # Full month name
print("Day (%d):", today.strftime("%d"))           # Day of the month (01-31)
print("Year (%Y):", today.strftime("%Y"))          # 4-digit year

# Combined format
print("Combined (%A, %B %d, %Y):", today.strftime("%A, %B %d, %Y"))
```

# Math in Python
Python provides basic math operations: +, -, *, /, //, %, **.

```python
a = 10
b = 3

# Addition
print("a + b =", a + b)
# Subtraction
print("a - b =", a - b)
# Multiplication
print("a * b =", a * b)
# Division
print("a / b =", a / b)
# Floor Division (removes decimals)
print("a // b =", a // b)
# Modulus (remainder)
print("a % b =", a % b)
# Exponentiation (power)
print("a ** b =", a ** b)
```

### 1. Built-in Math Functions
Python has several built-in functions:

```python
x = -4.7
print("Absolute value:", abs(x))
print("Round value:", round(x))
print("Min & Max:", min(5, 12, 3), max(5, 12, 3))
```

### 2. The Math Module
The math module provides more advanced mathematical functions.

```python
import math

# Square root
print("sqrt(16):", math.sqrt(16))

# Rounding
print("ceil(4.2):", math.ceil(4.2))   # round up
print("floor(4.9):", math.floor(4.9)) # round down

# Factorial
print("factorial(5):", math.factorial(5))

# Constants
print("pi:", math.pi)
print("e:", math.e)

# Power & Log
print("pow(2, 5):", math.pow(2, 5))   # 2^5
print("log10(100):", math.log10(100)) # base-10 log

# Trigonometry (degrees → radians)
print("sin(90°):", math.sin(math.radians(90)))
print("cos(0°):", math.cos(math.radians(0)))
```

# JSON in Python
JSON (JavaScript Object Notation) is a lightweight data format. Python can parse JSON and convert it to Python objects and vice versa.

### 1. Parse JSON - Convert from JSON to Python
Use json.loads() to convert JSON string to Python dictionary:

```python
import json  # Import the JSON module to work with JSON data

# JSON string
json_data = '{"name": "Alice", "age": 25, "city": "London"}'

# Convert JSON string to Python dictionary
python_dict = json.loads(json_data)

print(python_dict)  # Print the Python dictionary: {'name': 'Alice', 'age': 25, 'city': 'London'}
print("Name:", python_dict['name'])  # Access and print the value of the 'name' key: Alice
```

### 2. Convert from Python to JSON
Use json.dumps() to convert Python objects to JSON strings:

```python
python_obj = {"name": "Bob", "age": 30, "city": "New York"}   # A normal Python dictionary (Python object)
json_str = json.dumps(python_obj)   # Convert the Python dictionary into a JSON-formatted string

print(json_str)   # Print the JSON string (this is the format used for APIs, files, and data exchange)
```

### 3. Format the Result
You can format JSON strings for readability:

```python
pretty_json = json.dumps(python_obj, indent=4)
print(pretty_json)
```

### 4. Order the Result
Sort JSON keys alphabetically:

```python
sorted_json = json.dumps(python_obj, sort_keys=True, indent=4)
print(sorted_json)
```

# Regex in Python
Regular expressions (RegEx) allow you to match patterns in text efficiently.

### 1. RegEx Module
Use the re module for all regex operations:

```python
import re
```

### 2. Metacharacters
Symbols with special meanings:

| Metacharacter | Meaning                     |
|---------------|----------------------------|
| .             | Any character except newline |
| ^             | Start of string             |
| $             | End of string               |
| *             | 0 or more repetitions       |
| +             | 1 or more repetitions       |
| ?             | 0 or 1 repetition           |
| {n}           | Exactly n repetitions       |


```python
# 1. Dot (.)
text = "cat bat rat"
pattern = r".at"  # Matches any character followed by 'at'
print(re.findall(pattern, text))  # ['cat', 'bat', 'rat']

# 2. Caret (^)
text = "Hello world"
pattern = r"^Hello"  # Matches 'Hello' only at the start
print(re.findall(pattern, text))  # ['Hello']

# 3. Dollar ($)
text = "Hello world"
pattern = r"world$"  # Matches 'world' only at the end
print(re.findall(pattern, text))  # ['world']

# 4. Asterisk (*)
text = "aaa ab a"
pattern = r"a*"  # Matches 0 or more 'a's
print(re.findall(pattern, text))  # ['aaa', 'a', '', 'a', '']

# 5. Plus (+)
text = "aaa ab a"
pattern = r"a+"  # Matches 1 or more 'a's
print(re.findall(pattern, text))  # ['aaa', 'a', 'a']

# 6. Question mark (?)
text = "color colour colr"
pattern = r"colou?r"  # Matches 'color' or 'colour'
print(re.findall(pattern, text))  # ['color', 'colour']

# 7. Curly braces {n}
text = "aa aaa aaaa"
pattern = r"a{3}"  # Matches exactly 3 'a's
print(re.findall(pattern, text))  # ['aaa', 'aaa']
```

### 3. Flags
Flags modify matching behavior:

```python
pattern = re.compile("python", re.IGNORECASE)  # case-insensitive
```

### 4. Special Sequences


Sequence    Meaning

### Special Sequences in Regex
| Sequence | Meaning |
|----------|---------|
| \d       | Digit (0-9) |
| \D       | Non-digit |
| \s       | Whitespace |
| \S       | Non-whitespace |
| \w       | Alphanumeric |
| \W       | Non-alphanumeric |


```python
text = "My phone number is 123-456-7890, email: user@example.com"

# 1. \d → Digit (0-9)
digits = re.findall(r"\d", text)
print("Digits:", digits)  
# Output: ['1', '2', '3', '4', '5', '6', '7', '8', '9', '0']

# 2. \D → Non-digit
non_digits = re.findall(r"\D", text)
print("Non-digits:", non_digits)  
# Output: ['M', 'y', ' ', 'p', 'h', 'o', ...] (all characters except digits)

# 3. \s → Whitespace
spaces = re.findall(r"\s", text)
print("Whitespace:", spaces)  
# Output: [' ', ' ', ' ', ' ', ' '] (all spaces)

# 4. \S → Non-whitespace
non_spaces = re.findall(r"\S", text)
print("Non-whitespace:", non_spaces)  
# Output: ['M', 'y', 'p', 'h', 'o', 'n', 'e', ...] (all non-space characters)

# 5. \w → Alphanumeric (letters, digits, underscore)
alphanum = re.findall(r"\w", text)
print("Alphanumeric:", alphanum)  
# Output: ['M', 'y', 'p', 'h', 'o', 'n', 'e', 'n', 'u', 'm', 'b', 'e', 'r', 'i', 's', '1', '2', ...]

# 6. \W → Non-alphanumeric (everything else)
non_alphanum = re.findall(r"\W", text)
print("Non-alphanumeric:", non_alphanum)  
# Output: [' ', ' ', ' ', '-', '-', '-', ',', ' ', ':', '@', '.', ...]
```

### 5. RegEx Functions
re.findall() → find all matches

re.search() → search for first match

re.split() → split string by pattern

re.sub() → replace pattern with string

### 6. The findall() Function
Returns all matches:

```python
text = "My phone number is 123-456-7890"
numbers = re.findall(r"\d+", text)
print(numbers)  # ['123', '456', '7890']
```

### 7. The search() Function
Returns first match as a Match object:

```python
match = re.search(r"\d+", text)
if match:
    print("First number:", match.group())
```

### 8. The split() Function
Splits text by pattern:

```python
parts = re.split(r"\s+", "Python is awesome")
print(parts)  # ['Python', 'is', 'awesome']
```

### 9. The sub() Function
Replace pattern with string:

```python
new_text = re.sub(r"\d+", "#", text)
print(new_text)  # My phone number is #-#-#
```

### 10. Match Object
Provides info about matched text:

```python
match = re.search(r"\d+", text)
print("Matched text:", match.group())
print("Start index:", match.start())
print("End index:", match.end())
```
