# While Loops in Python
In this lecture, we learn how Python uses **while loops** to repeat actions until a condition becomes false. We will also cover infinite loops, common mistakes, and examples using `break`, `continue`, `for`, and nested loops.

## What is a While Loop?
A `while` loop repeats a block of code **as long as its condition ramins True**.<br>
&nbsp;&nbsp;&nbsp;`while` something is **true**, repeat

**Syntax:**
```python
while condition:
    # code block
```

````python
count = 1
while count <= 5:
    print(count)
    count+=1
````

## How While Loops Operate Step-by-Step
1. Python checks the condition.
2. If it's `True`, it runs the loop body.
3. After finishing the body, Python starts over from step 1.
4. When the condition becomes `False`, the loop stops.

## Example 1: Summing Numbers 1 to 10

````python
total = 0
num = 1
while num <= 10:
    total+=num
    num+=1
print(total)
````

## Using break in While Loops
`break` stops the loop immediately.

````python
x = 1
while x <= 10:
    if x == 5:
        break
    print(x)
    x+=1
````

## Using continue in While Loops
`continue` skips the current iteration and moves to the next one.

````python
x = 0
while x < 5:
    x+=1
    if x == 3:
        continue
    print(x)
````

## Example 2: Security System Simulation
User has 3 attempts to enter the correct password.

````python
password = "python123"
attempts = 3
while attempts > 0:
    entry = input("Enter password: ")
    if entry == password:
        print("Access Granted.")
        break
    attempts-=1
    print(f"Wrong password. {attempts} attempts left.")
if attempts == 0:
    print("Access Denied.")
````

## Example 3: Loading Bar Simulation 
This shows how loops can simulate progress bars.

````python
import time
progress = 0
while progress <= 100:
    print(f"Loading... {progress}%", end='\r')
    time.sleep(0.5)
    progress+=5
print("\nDone!")
````

## Example 4: Prime Number Checker Using While Loop
Determining if a number is prime using `while` and break.

````python
n = 29
is_prime= True
i = 2
while i * i <= n:
    if n % i == 0:
        is_prime = False
        break
    i+=1
print(n, "is prime?", is_prime)
````

## Comparison: While vs For Loops
Both can repeat code, but their usage is different:
- Use `for` when iterating over a sequence (list, string, range).
- Use `while` when repeating until a condition changes.

````python
# for loop :
for i in range(5):
    print(i)
````

````python
# doing the same with while loop : 
i = 0
while i < 5:
    print(i)
    i+=1
````

## Nested Loops
You can place loops inside loops.
Here we use both `for` and `while`.

````python
i = 1
while i <= 3:
    print('Outer loop iteration:', i)
    for j in range(1, 4):
        print('   Inner loop value:', j)
    i+=1
````

## Nested While Loop Example: Multiplication Table (1 to 5)

````python
i = 1
while i <= 5:
    j = 1
    while j <= 5:
        print(i * j, end=' ')
        j+=1
    print()
    i+=1
````

# Eample 5: while-else
searching for an item in a list

````python

numbers = [10, 52, 3, 74, 18, 9, 41]
target_value = 9
index = 0

while index < len(numbers):
    if numbers[index] == target_value:
        print(f"found {target_value} at index {index}")
        break  
    index+=1
else:
    print(f"{target_value} was not found in the list")
````
