## For Loops

A **for loop** in Python runs a block of code repeatedly, once for each
item in a sequence. The basic syntax is:



```python
for item in sequence:
    do_something(item)
```



Example:



```python
for i in range(1, 5):
    print("Hello!", i)
```



## Looping Through a String

Strings are sequences of characters. You can loop through each character
with a for loop:



```python
word = "Python"
for letter in word:
    print(letter)
```



## The `range()` Function

`range()` generates numbers for looping.

-   `range(stop)`
-   `range(start, stop)`
-   `range(start, stop, step)`

Examples:



```python
for i in range(5):
    print(i)
```

```python
for i in range(2, 6):
    print(i)
```

```python
for i in range(2, 10, 3):
    print(i)
```



## `break`, `continue`, and `pass`

### `break`

Exits a loop immediately.



```python
for num in range(1, 6):
    print("Checking:", num)
    if num == 3:
        print("Found 3! Breaking out.")
        break
print("Loop is done.")
```



### `continue`

Skips the rest of the current iteration.



```python
for num in range(1, 6):
    if num == 3:
        print("Skipping", num)
        continue
    print("Number is", num)
```



### `pass`

Does nothing; used as a placeholder.



```python
for i in range(3):
    pass
print("Loop finished; i is still", i)
```



## else in For Loops

The `else` block runs only if the loop finishes normally (no break).



```python
for c in "abc":
    print(c)
else:
    print("Done looping through characters.")
```



If a break occurs, else is skipped:



```python
for c in "abc":
    if c == "b":
        print("Found b, breaking.")
        break
    print(c)
else:
    print("Loop completed without break.")
```



## Nested for Loops

A loop inside another loop.



```python
letters = "AB"
numbers = "12"
for letter in letters:
    for number in numbers:
        print(letter + number)
```
